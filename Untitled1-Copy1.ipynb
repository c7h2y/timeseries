```python
import numpy as np
import pandas as pd

def generate_sequences(
    data: pd.DataFrame,
    target_col: str,
    sampling_rate: int = 6,
    sequence_length: int = 120,
    delay: int = None
):
    """
    data: pandas DataFrame containing features and target_col
    target_col: 予測したい値を含む列名
    sampling_rate: サンプリング間隔（行数）
    sequence_length: １入力シーケンスの長さ（サンプル数）
    delay:  予測ターゲットのオフセット。Noneの場合、以下の式で自動計算。
            delay = sampling_rate + (sequence_length + 24 - 1)
    
    return:
      X: np.ndarray of shape (サンプル数, sequence_length, 特徴量数)
      y: np.ndarray of shape (サンプル数,)
    """
    if delay is None:
        delay = sampling_rate + (sequence_length + 24 - 1)

    values = data.values
    targets = data[target_col].values
    n_rows, n_features = values.shape

    # シーケンス末尾とターゲット位置がデータ範囲内に収まる最終開始位置
    max_start = n_rows - max((sequence_length - 1) * sampling_rate, delay) - 1

    X, y = [], []
    for start in range(max_start + 1):
        # sequence_length 個のサンプルを sampling_rate ごとに取得
        idx = start + np.arange(sequence_length) * sampling_rate
        X.append(values[idx])
        # ターゲットは start + delay の行
        y.append(targets[start + delay])

    X = np.stack(X)            # shape=(サンプル数, sequence_length, n_features)
    y = np.array(y)            # shape=(サンプル数,)
    return X, y

# -------------------------
# 使用例
# -------------------------
df = pd.read_csv("jena_climate_2009_2016.csv")
X, y = generate_sequences(df, target_col="T (degC)")
print(X.shape, y.shape)
```
